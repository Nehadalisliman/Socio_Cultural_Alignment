```python
# اختبار استجابة النموذج لبيانات وثقافات مختلفة
prompt_1 = "In our culture, family traditions emphasize"
prompt_2 = "When greeting elders in a traditional community,"

# إدخال الجمل للنموذج
input_ids_1 = tokenizer(prompt_1, return_tensors="pt")
input_ids_2 = tokenizer(prompt_2, return_tensors="pt")

# توليد الإجابة للقول الأول
output_1 = model.generate(**input_ids_1, max_length=30, num_return_sequences=1)
print("--- تجربة الجملة الأولى ---")
print(tokenizer.decode(output_1[0], skip_special_tokens=True))

# توليد الإجابة للقول الثاني
output_2 = model.generate(**input_ids_2, max_length=30, num_return_sequences=1)
print("\n--- تجربة الجملة الثانية ---")
print(tokenizer.decode(output_2[0], skip_special_tokens=True))
```

```python
# 1. إعداد مجموعة بيانات ثقافية مصغرة (Mini Cultural Dataset)
cultural_dataset = [
    {
        "prompt": "In our culture, family traditions emphasize",
        "completion": " mutual support, respect for elders, and preserving core community values across generations."
    },
    {
        "prompt": "When greeting elders in a traditional community,",
        "completion": " it is customary to show deep respect, offer polite greetings, and listen attentively to their wisdom."
    },
    {
        "prompt": "Hospitality in traditional societies is seen as",
        "completion": " a fundamental moral duty, where guests are welcomed with generous food, warmth, and honor."
    }
]

print("تم تجهيز عينة البيانات الثقافية بنجاح!")
print(f"عدد الأمثلة: {len(cultural_dataset)}")

# 2. إرجاع النماذج وتوليد الاستجابة المحسّنة (Simulation/Evaluation)
print("\n--- مقارنة الاستجابة الموجهة ثقافياً ---")
for item in cultural_dataset:
    print(f"\n[الprompt]: {item['prompt']}")
    print(f"[النص المكتمل والموجه ثقافياً]: {item['prompt']}{item['completion']}")
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

# سنستخدم نموذج خفيف وسريع للتجربة
model_name = "distilgpt2"

print("جاري تحميل النموذج...")
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

print("تم تحميل النموذج بنجاح!")

# اختبار بسيط للنص
inputs = tokenizer("Hello, AI is", return_tensors="pt")
outputs = model.generate(**inputs, max_length=20)
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```
