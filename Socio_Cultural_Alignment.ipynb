```python
# 1. تثبيت المكتبات المطلوبة
!pip install -q trl transformers datasets accelerate torch

import torch
from transformers import AutoModelForCausalLM, AutoTokenizer
from datasets import Dataset
from trl import DPOTrainer, DPOConfig

print("تم تثبيت وتجهيز مكتبات الـ DPO بنجاح!")
```

```python
# 2. إعداد داتاسيت التفضيلات الثقافية (Cultural Preference Dataset)
train_data = {
    "prompt": [
        "In our culture, family traditions emphasize",
        "When greeting elders in a traditional community,",
        "Hospitality in traditional societies is seen as"
    ],
    "chosen": [
        "mutual support, respect for elders, and preserving core community values across generations.",
        "it is customary to show deep respect, offer polite greetings, and listen attentively to their wisdom.",
        "a fundamental moral duty, where guests are welcomed with generous food, warmth, and honor."
    ],
    "rejected": [
        "the importance of family and family. We are not just about family, we are about family.",
        "the elders are often asked to give a greeting to the elders.",
        "hospitality and hospitality. People like food and food in society."
    ]
}

dataset = Dataset.from_dict(train_data)

# 3. تحميل النموذج والـ Tokenizer (distilgpt2)
model_id = "distilgpt2"
model = AutoModelForCausalLM.from_pretrained(model_id)
model_ref = AutoModelForCausalLM.from_pretrained(model_id)
tokenizer = AutoTokenizer.from_pretrained(model_id)

# إعداد الـ Pad Token إذا لم يكن موجوداً
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

print("تم تجهيز البيانات وتحميل نموذج distilgpt2 بنجاح!")
```

```python
# 4. إعداد معاملات الـ DPO وبدء التدريب
training_args = DPOConfig(
    output_dir="./dpo_cultural_output",
    learning_rate=1e-5,
    per_device_train_batch_size=1,
    gradient_accumulation_steps=2,
    num_train_epochs=3,
    logging_steps=1,
    save_strategy="no",
    report_to="none"
)

dpo_trainer = DPOTrainer(
    model=model,
    ref_model=model_ref,
    args=training_args,
    train_dataset=dataset,
    tokenizer=tokenizer,
)

# 5. تنفيذ عملية التدريب الفعلي
print("جاري بدء التدريب (DPO Training)...")
dpo_trainer.train()
print("تم التدريب بنجاح تام وإتمام محاذاة النموذج!")
```

```python
import torch
import torch.nn.functional as F

print("🚀 جاري محاكاة وتطبيق خطوة الـ DPO Alignment يدوياً لضمان الاستقرار...")

# 1. حساب الخسارة (Loss) لكل من الإجابة المفضلة chosen والمرفوضة rejected
model.train()
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5)

for i in range(len(dataset)):
    prompt = dataset[i]["prompt"]
    chosen = dataset[i]["chosen"]
    rejected = dataset[i]["rejected"]

    # ترميز المدخلات
    inputs_chosen = tokenizer(prompt + " " + chosen, return_tensors="pt", truncation=True, max_length=64)
    inputs_rejected = tokenizer(prompt + " " + rejected, return_tensors="pt", truncation=True, max_length=64)

    # حساب الـ Logits للنموذج
    outputs_chosen = model(**inputs_chosen, labels=inputs_chosen["input_ids"])
    outputs_rejected = model(**inputs_rejected, labels=inputs_rejected["input_ids"])

    loss_chosen = outputs_chosen.loss
    loss_rejected = outputs_rejected.loss

    # دالة هدف الـ DPO المبسطة (تعظيم احتمال Chosen وتقليل احتمال Rejected)
    dpo_loss = -F.logsigmoid(loss_rejected - loss_chosen)

    optimizer.zero_grad()
    dpo_loss.backward()
    optimizer.step()

print(f"✅ تم الانتهاء من دورات تدريب ومحاذاة الـ DPO بنجاح! Loss الأخير: {dpo_loss.item():.4f}")
```
